Part A — build each core plot type

Task 01

Line and scatter

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 4 * np.pi, 200)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, np.sin(x), label = "sin(x)", linewidth = 2)
ax.plot(x, np.cos(x), label = "cos(x)", linewidth = 2)

ax.set_title("Sin and Cos waves")
ax.set_xlabel("X (radians)")
ax.set_ylabel("Amplitude")
ax.legend()
ax.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)

x = np.random.rand(100) * 10
noise = np.random.rand(100) * 2
y = 2 * x + noise

fig, ax = plt.subplots(figsize=(8, 5))

ax.scatter(x, y)

ax.set_title("Scatter Plot : 2x + Noise.")
ax.set_xlabel("X (Units)")
ax.set_ylabel("Y (Units)")
ax.grid(True)
plt.show()
```

Rule :
Use the line plot when data has a continuous order (time, sequence, functions).
Example (ML) : Training loss vs Epochs

Use the scatter plot when showing the relation between two variables or individual observation.
Example (ML) : Predict values vs Actual values or Height vs weight dataset.

```python
import matplotlib.pyplot as plt
import numpy as np

color = np.random.rand(100)
size = np.random.rand(100) * 300

fig, ax = plt.subplots(figsize=(8, 5))

scatter = ax.scatter(x, y, c=color, s=size, cmap="viridis", alpha=0.7)

ax.set_title("Enhanced Scattered Plot")
ax.set_xlabel("X (Units)")
ax.set_ylabel("Y (Units)")
ax.grid(True)
plt.show()
```

Task 02

Bar chart versus histogram

```python
import matplotlib.pyplot as plt

categories = ["North", "South", "East", "West", "Central"]
sales = [120, 130, 150, 100, 180]

fig, ax = plt.subplots(figsize=(6, 4))

ax.bar(categories, sales, color="skyblue")
ax.set_title("Sales by Regions")
ax.set_xlabel("Categories")
ax.set_ylabel("sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(42)
data = np.random.normal(50, 10, 1000)

fig, ax = plt.subplots(figsize=(6, 4))

ax.hist(data, bins = 30, color = "lightgreen", edgecolor = "black")

mean = np.mean(data)
ax.axvline(mean, color="blue", linestyle="--", linewidth=2, label=f"Mean = {mean:.3f}")

ax.set_title("Histogram of normally distributed data")
ax.set_xlabel("Value")
ax.set_ylabel("Frequency")
ax.legend()

plt.show()
```

Bar charts has gaps because each bar represents a seperate category.

Histograms doesn't has gaps because it represents continuous numerical data, 
where adjacent intervals (bins) touch each other.

```python
import matplotlib.pyplot as plt
import numpy as np

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
bin_list = [5, 30, 200]

for ax, bins in zip(axes, bin_list):
    ax.hist(data, bins=bins, color="orange", edgecolor="white")
    ax.axvline(mean, color="blue", linestyle="--", linewidth=2)
    ax.set_title(f"Bins={bins}")

plt.tight_layout()
plt.show()
```

Task 03

Subplots — a grid that tells one story

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

np.random.seed(42)
scatter_x = np.random.rand(50)
scatter_y = np.random.rand(50)

categories = ["A", "B", "C", "D", "E"]
values = [15, 20, 25, 32, 44]

hist_data = np.random.randn(1000)

fig, ax = plt.subplots(2, 2, figsize=(10, 8))

# A line plot :-

ax[0, 0].plot(x, y, color="blue")
ax[0, 0].set_title("Line Plot")
# ax[0, 0].grid(True)

# A scatter plot :-

ax[0, 1].scatter(scatter_x, scatter_y, color="red")
ax[0, 1].set_title("Scatter Plot")
# ax[0, 1].grid(True)

# A bar chart :-

ax[1, 0].bar(categories, values, color="Green", edgecolor="black")
ax[1, 0].set_title("Bar Chart")
# ax[1, 0].grid(True)

# A histogram :- 

ax[1,1].hist(hist_data, bins=30, color="orange", edgecolor="white")
ax[1,1].set_title("Histogram")
# ax[1,1].grid(True)

# Whole figure title :-

fig.suptitle("2x2 Grid of Different Plots", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)

fig, ax = plt.subplots(1, 3, figsize=(12, 4))

# Line plot :-

ax[0].plot(x, np.sin(x))
ax[0].set_title("Line Plot")
# ax[0].grid(True)

# Scattered plot :-

ax[1].scatter(np.random.rand(50), np.random.rand(50))
ax[1].set_title("Scattered Plot")
# ax[1].grid(True)

# Bar Chart :-

# categories = ["A", "B", "C", "D", "E"]
# values = [150, 180, 142, 123, 139]

# ax[2].bar(categories, values)
# ax[2].set_title("Bar Chart")
# ax[2].grid(True)

# Histogram :-

ax[2].hist(np.random.randn(500), bins = 20, edgecolor = "white")
ax[2].set_title("Histogram")
# ax[2].grid(True)

fig.suptitle("1x3 Grid")

plt.tight_layout()
plt.show()
```

Part B — judgment, not just mechanics

Task 04

Fix the terrible chart

```python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.bar(['A', 'B', 'C'], [102, 104, 106], color = ['red', 'lime', 'cyan'])
ax.set_ylim(100, 107)
plt.show()
```

1. No chart title
2. No x axis label
3. No y axis label
4. y axis starts with 100 instead of 0
5. No gridlines, making comparision harder
6. No legend
7. Clashing, distracting colours(red, lime, cyan)

```python
import matplotlib.pyplot as plt

categories = ['A', 'B', 'C']
values = [102, 104, 106]

fig, ax = plt.subplots(figsize=(6, 4))

ax.bar(categories, values, color='teal', label='values')

ax.set_title("Comparision of Categories")
ax.set_xlabel("Categories")
ax.set_ylabel("Values")

ax.set_ylim(0, 110)

ax.grid(axis='y', linestyle='--', alpha=0.6)

ax.legend()

plt.tight_layout()
plt.show()
```

Task 05

Choose the right plot

```python
# a. How has our monthly revenue changed over the past two years?
# ans. Line chart

import numpy as np
import matplotlib.pyplot as plt

months = np.arange(1, 25)
revenue = [50, 52, 55, 53, 57, 60, 62, 64, 63, 66, 68, 70,
           72, 74, 73, 76, 78, 80, 82, 84, 85, 87, 89, 92]

fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(months, revenue, marker='o')
ax.set_title("Monthly Revenue Over Two Years")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue (1000/- RS)")
ax.grid(True)
plt.show()
```

```python
# b. Is there a relationship between a house's size and its price?
# ans. Scattered Plot

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

house_size = np.random.randint(800, 3000, 100)
price = house_size * 180 + np.random.normal(0, 50000, 100)

fig, ax = plt.subplots(figsize=(6, 5))

ax.scatter(house_size, price, alpha=0.7)
ax.set_title("House Size vs Price")
ax.set_xlabel("House Size (sq ft)")
ax.set_ylabel("Price (Rs.)")
ax.grid(True)
plt.show()
```

```python
# c. How are exam scores distributed across the class — clustered or spread?
# ans. Histogram

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(2)

scores = np.random.normal(70, 12, 200)

fig, ax = plt.subplots(figsize=(6, 5))

ax.hist(scores, bins=15, color="gray", edgecolor="black")
ax.set_title("Distribution of Exam Scores")
ax.set_xlabel("Score")
ax.set_ylabel("Number of Students")
ax.grid(True)
plt.show()
```

```python
# d. Which of our five products sells the most?
# ans. Bar Chart

import matplotlib.pyplot as plt

products = ['A', 'B', 'C', 'D', 'E']
sales = [452, 520, 342, 457, 825]

fig, ax = plt.subplots(figsize=(6, 4))

ax.bar(products, sales, color = "crimson")

ax.set_title("Product Sales")
ax.set_xlabel("Products")
ax.set_ylabel("Unit Sold")
ax.grid(axis="y")
plt.show()
```

```python
# e. How do the salary distributions compare across four departments?
# ans. Box Plot

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(3)

it = np.random.normal(65000, 7000, 50)
hr = np.random.normal(40000, 5000, 50)
sales = np.random.normal(55000, 5500, 50)
finance = np.random.normal(60000, 6000, 50)

fig, ax = plt.subplots(figsize=(7, 5))

ax.boxplot([it, hr, sales, finance],
          label = ["IT", "HR", "Sales", "Finance"])

ax.set_title("Salary Distribution by Department")
ax.set_xlabel("Departments")
ax.set_ylabel("Salary (Rs.)")
ax.grid(axis="y", linestyle = "--", alpha= 0.7)
plt.show()
```

```python
# f. What proportion of our users are on each of three subscription tiers?
# ans. Pie Chart

import matplotlib.pyplot as plt

tiers = ["Basic", "Standard", "Premium"]
users = [450, 320, 230]

fig, ax = plt.subplots(figsize=(6, 6))

ax.pie(users, labels = tiers, autopct = "%1.1f%%", startangle = 90)
ax.set_title("Subscription Tier Distribution")
plt.show()
```

Part C — the polished figure

Task 06

One report-quality figure

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate Dataset of 200 Students
np.random.seed(42)

students = 200

study_hours = np.random.uniform(1, 10, students)
attendance = np.random.randint(50, 101, students)
exam_score = (study_hours * 6 + attendance * 0.4 + np.random.normal(0, 8, students))
exam_score = np.clip(exam_score, 0, 100)

# Passed / Failed
passed = exam_score >= 50
colors = np.where(passed, "green", "red")

# Create Figure
fig, axes = plt.subplots(2, 2, figsize=(12, 9))


# The relationship, coloured by pass/fail
# Penal 1 scatter plot
axes[0, 0].scatter(study_hours[passed], exam_score[passed], color="green", alpha=0.6, label="passed")
axes[0, 0].scatter(study_hours[~passed], exam_score[~passed], color="red", alpha=0.6, label="failed")
axes[0, 0].axhline(50, color="black", linestyle="--", label="pass mark")
axes[0, 0].set_title("Study Hours vs Exam Score")
axes[0, 0].set_xlabel("Hours Studied")
axes[0, 0].set_ylabel("Exam Score")
axes[0, 0].legend()


# Distribution of scores
# Penal 2 Histogram
axes[0, 1].hist(exam_score, bins=25, color="springgreen", edgecolor="white")
axes[0, 1].axvline(50, color="black", linestyle="--", linewidth=2)
axes[0, 1].set_title("Distribution of Scores")
axes[0, 1].set_xlabel("Exam Score")
axes[0, 1].set_ylabel("Count")


# Pass rate by hours-studied bucket
# Penal 3 Bar Chart 
buckets = np.digitize(study_hours, [2, 4, 6, 8])
labels = ["0-2", "2-4", "4-6", "6-8", "8-10"]
rates = [passed[buckets==i].mean() * 100 if (buckets == i).any() else 0 for i in range(5)]
axes[1, 0].bar(labels, rates, color="orange")
axes[1, 0].set_title("Pass rate by hours studied")
axes[1, 0].set_xlabel("Hours studied")
axes[1, 0].set_ylabel("pass rate(%)")


# How many passed vs failed
# Penal 4 Bar Chart
axes[1, 1].bar(["passed", "failed"], [passed.sum(), (~passed).sum()], color=["green", "red"])
axes[1, 1].set_title("Overall Outcome")
axes[1, 1].set_ylabel("Number of Students")

fig.suptitle("Does studying help? A four-penal answer.", fontsize=16, fontweight="bold")

plt.tight_layout()
plt.show()
```
