```python
import numpy as np
import pandas as pd

df = pd.read_csv("titanic.csv")
print(f"shape: {df.shape[0]}, passengers, {df.shape[1]} columns\n")
df.head()
```

```python
import numpy as np
import pandas as pd

# Total Passenger
print(f"Total Passenger : {len(df)}")
print(f"Total Passenger : {df.shape[0]}")

# What fraction of passengers survived overall?
survival_rate = df["survived"].mean()
print(f"\nOverall Survival Rate : {survival_rate:.4f}")
print(f"Overall Survival Percentage : {survival_rate * 100:.2f}%")

# What was the survival rate for women versus men?
print(f"\nFemale Survival Rate  : {df[df["sex"] == "female"]["survived"].mean():.4f}")
print(f"Male Survival Rate : {df[df["sex"] == "male"]["survived"].mean():.4f}")

# What was the survival rate for each passenger class (1st, 2nd, 3rd)?
print(f"\n1st Class Survival Rate : {df[df['pclass'] == 1]['survived'].mean():.3f}")
print(f"2nd Class Survival Rate : {df[df['pclass'] == 2]['survived'].mean():.3f}")
print(f"3rd Class Survival Rate : {df[df['pclass'] == 3]['survived'].mean():.3f}")

# What was the average age of passengers?
average_age = df["age"].mean()
print("\nAverage Age:", average_age)

# Who was the oldest passenger, and who was the youngest?
print(f"\nOldest One: {df["age"].max()}")
print(f"Youngest One: {df["age"].min()}")

# What was the average fare paid in each class?
print(f"\nAverage Fare in 1st class : {df[df['pclass'] == 1]['fare'].mean():.3f}")
print(f"Average Fare in 2nd class : {df[df['pclass'] == 2]['fare'].mean():.3f}")
print(f"Average Fare in 3rd class : {df[df['pclass'] == 3]['fare'].mean():.3f}")

# How many passengers were children (under 18)?
children = df[df["age"] < 18]
print(f"\nNumber of Children : {len(children)}")

# What was the survival rate among children?
print(f"\nSurvival rate of Children : {children['survived'].mean():.3f}")

# How many passengers embarked from each port?
print(f"\nS : {(df['embarked'] == 'S').sum()}")
print(f"C : {(df['embarked'] == 'C').sum()}")
print(f"Q : {(df['embarked'] == 'Q').sum()}")

# How many passengers have a missing age?
print(f"\nMissing Age : {df['age'].isna().sum()}")

# What was the survival rate for women in 1st class?
print(f"\nWomen in First class survival rate : {df[(df['sex'] == 'female') & (df['pclass'] == 1)]['survived'].mean():.4f}")

# What was the survival rate for men in 3rd class?
print(f"\nMen in Third class survival rate : {df[(df['sex'] == 'male') & (df['pclass'] == 3)]['survived'].mean():.4f}")

# What was the average family size on board (siblings/spouses + parents/children)?
print(f"\nAverage Family Size : {(df['sibsp'] + df['parch']).mean():.3f}")

# Did passengers travelling alone survive at a different rate than those with family?
print(f"\nSurvival rate for passengers travelling alone: {df[(df['sibsp'] + df['parch']) == 0]['survived'].mean():.4f}")
print(f"Survival rate for passengers travelling with family: {df[(df['sibsp'] + df['parch']) > 0]['survived'].mean():.4f}")
```

Part B — handle missing age three ways (+ reflection)

TASK 02 

Three strategies for missing age

The curriculum's missing-data activity, and the reflection that goes with the deliverable.

```python
# Confirm how many age values are missing, and what fraction of the data that is
missing_age = df['age'].isna().sum()
missing_fraction = missing_age / len(df)
total_rows = len(df)

print(f"Missing age : {missing_age}")
print(f"Missing Fraction : {missing_fraction:.4f}")
print(f"Percentage Missing : {missing_fraction * 100:.4f}")

# Strategy 1 — Drop:
dropped = df.dropna(subset = ['age'])
print("\n" + "=" * 16)
print("Strategy:1 DROP")
print("=" * 16)
print("Rows Remaining:", len(dropped))
print(f"Mean Age: {dropped['age'].mean():.2f}")
print(f"Std Age: {dropped['age'].std():.1f}")

# Strategy 2 — Mean fill:
filled_mean = df.copy()
filled_mean['age'] = filled_mean['age'].fillna(df['age'].mean())

print("\n" + "=" * 21)
print("Strategy:2 MEAN FILL")
print("=" * 21)
print(f"Rows Remaining: {len(filled_mean)}")
print(f"Mean Age: {filled_mean['age'].mean():.2f}")
print(f"Std Age: {filled_mean['age'].std():.1f}")

# Strategy 3 — Group fill:
filled_group = df.copy()
filled_group['age'] = filled_group.groupby(['pclass', 'sex'])['age'].transform(lambda x: x.fillna(x.median()))

print("\n" + "=" * 21)
print("Strategy:3 GROUP FILL")
print("=" * 21)
print(f"Rows Remaining: {len(filled_group)}")
print(f"Mean Age: {filled_group['age'].mean():.2f}")
print(f"Std Age: {filled_group['age'].std():.1f}")

print(
    '''\n\nGroup median is smarter because it uses information from
the passenger's class and sex, giving a more realistic age 
estimate than using the same overall mean for everyone.
    '''
)


# Side by side comparision
print("\n" + "=" * 25)
print("Coparision side by side")
print("=" * 25)

print("Strategy           Mean Age          Std Age")
print(f"Drop                {dropped['age'].mean():.2f}             {dropped['age'].std():.1f}")
print(f"Mean Fill           {filled_mean['age'].mean():.2f}             {filled_mean['age'].std():.1f}")
print(f"Group Mean          {filled_group['age'].mean():.2f}             {filled_group['age'].std():.1f}")


# Women vs Men survival rate
print("\n" + "=" * 16)
print("Strategy:1 DROP")
print("=" * 16)

print(f"Women: {dropped[dropped['sex'] == 'female']['survived'].mean():.3f}")
print(f"Men: {dropped[dropped['sex'] == 'male']['survived'].mean():.3f}")

print("\n" + "=" * 21)
print("Strategy:2 MEAN FILL")
print("=" * 21)

print(f"Women: {filled_mean[filled_mean['sex'] == 'female']['survived'].mean():.3f}")
print(f"Men: {filled_mean[filled_mean['sex'] == 'male']['survived'].mean():.3f}")

print("\n" + "=" * 21)
print("Strategy:3 GROUP FILL")
print("=" * 21)

print(f"Women: {filled_group[filled_group['sex'] == 'female']['survived'].mean():.3f}")
print(f"Men: {filled_group[filled_group['sex'] == 'male']['survived'].mean():.3f}")
```

TASK 03 

Reflection: which strategy, and why?

The written reflection that completes the deliverable. This is about judgement, not code.

The three missing-age strategies each have different advantages and disadvantages. Dropping rows with missing age is simple and does not change the distribution of the remaining age values, but it reduces the amount of data available for analysis. Filling missing ages with the overall mean keeps all rows, but it reduces the standard deviation because every missing value is replaced by the same middle value. Group-based filling is usually better because it uses information from groups such as passenger class or sex, so the imputed ages can better reflect differences between those groups. For this Titanic dataset, I would choose group-based imputation because it keeps all the data while better preserving the patterns and spread between different passenger groups. However, there is no single “correct” strategy; the best choice depends on the goal—for example, if accurate individual ages were essential and only a small number of rows were missing, dropping those rows might be preferable. Transparency is also important for trust: if someone imputed 20% of a dataset without mentioning it, the analysis could be misleading because a large portion of the data would contain estimated rather than observed values, and others would not know how those values were created.

Part C — filter, sort, rank

TASK 04 

Multi-condition filters and ranking

The curriculum's filtering and sorting activity. Sharpen your selection skills.

```python
# Female 1st, 2nd class under 30
young_f = df[(df['sex'] == 'female') & (df['pclass'].isin([1, 2])) & (df['age'] < 30)]

print("\n" + "=" * 40)
print("Females in 1st, 2nd class and under 30")
print("=" * 40)

print("Total Number:", len(young_f))
print(f"Survival Rate: {young_f['survived'].mean():.3f}")
print(f"Survival Rate in Percent: {young_f['survived'].mean() * 100:.2f}")


# 10 oldest passenger who survived
print("\n" + "=" * 35)
print("10 oldest passenger who survived")
print("=" * 35)

oldest_survivors = df[df['survived'] == 1].sort_values(
    'age', ascending=False
).head(10)
print("10 oldest survivors:")
print(oldest_survivors[['sex', 'pclass', 'age', 'fare']])


# Top 10% vs Bottom 10% fare payers
print("\n" + "=" * 35)
print("Top 10% vs Bottom 10% fare payers")
print("=" * 35)

top_10 = df[df['fare'] >= df['fare'].quantile(0.90)]
bottom_10 = df[df['fare'] <= df['fare'].quantile(0.10)]

print(f"Top 10% Survival Rate: {top_10['survived'].mean():.3f}")
print(f"Bottom 10% Survival Rate: {bottom_10['survived'].mean():.3f}")


# Sort by class ascending and fare descending
print("\n" + "=" * 45)
print("Sort by class ascending and fare descending")
print("=" * 45)

sorted_cf = df.sort_values(['pclass', 'fare'], ascending=[True, False])

print("Sorted by class and fare:", "\n", sorted_cf)


# Describe fare and age
print("\n" + "=" * 21)
print("Describe Fare and Age")
print("=" * 21)

print("Fare Discription")
print(df['fare'].describe())
print("\nAge Discription")
print(df['age'].describe())


# Value counts of Class and Embarkation
print("\n" + "=" * 38)
print("Value counts of Class and Embarkation")
print("=" * 38)

print('Passengers by class:')
print(df['pclass'].value_counts())
print("\nWhere they embarked")
print(df['embarked'].value_counts())
```
